1. Quantos ride_id estão duplicados?

```python
duplicados_id = df[df.duplicated(subset="ride_id", keep=False)]

print("Linhas envolvidas em IDs duplicados:", len(duplicados_id))
print("Quantidade de IDs duplicados:", duplicados_id["ride_id"].nunique())
```

4.Criando as durações das viagens


```python
df["ride_length_min"] = (
    (df["ended_at"] - df["started_at"]).dt.total_seconds() / 60
)
```

```python
df["ride_length_min"].describe()
```

5. Eliminando durações negativas


```python
print("Duração negativa:", (df["ride_length_min"] < 0).sum())
print("Duração igual a zero:", (df["ride_length_min"] == 0).sum())
print("Duração <= 0:", (df["ride_length_min"] <= 0).sum())
```

```python
df[df["ride_length_min"] <= 0][
    ["ride_id", "started_at", "ended_at", "ride_length_min", "member_casual"]
].sort_values("ride_length_min").head(30)
```

```python
#essas 29 viagens não são necessariamente viagens inválidas. É uma ambiguidade causada pelo horário de verão em Chicago, em 2025:
negativas = df[df["ride_length_min"] < 0]

print(negativas["started_at"].dt.date.value_counts())
print()
print(negativas["ride_length_min"].describe())
```

```python
#Removendo:
df = df[df["ride_length_min"] > 0].copy()

df.shape
```

6.Agregando

```python
df.groupby("member_casual")["ride_length_min"].agg(
    quantidade="count",
    media="mean",
    mediana="median",
    minimo="min",
    maximo="max"
)
```

```python
print("Viagens > 6h:", (df["ride_length_min"] > 360).sum())
print("Viagens > 12h:", (df["ride_length_min"] > 720).sum())
print("Viagens > 24h:", (df["ride_length_min"] > 1440).sum())
print("Maior duração:", df["ride_length_min"].max())
```

```python
df.nlargest(20, "ride_length_min")[
    [
        "ride_id",
        "started_at",
        "ended_at",
        "ride_length_min",
        "member_casual",
        "rideable_type"
    ]
]
```

```python
print("Viagens < 1 minuto:",
      (df["ride_length_min"] < 1).sum())

print("Viagens entre 1 e 60 minutos:",
      ((df["ride_length_min"] >= 1) &
       (df["ride_length_min"] <= 60)).sum())

print("Viagens entre 1h e 6h:",
      ((df["ride_length_min"] > 60) &
       (df["ride_length_min"] <= 360)).sum())

print("Viagens entre 6h e 24h:",
      ((df["ride_length_min"] > 360) &
       (df["ride_length_min"] <= 1440)).sum())

print("Viagens > 24h:",
      (df["ride_length_min"] > 1440).sum())
```

7.Entendendo se há padrões

```python
faixas = pd.cut(
    df["ride_length_min"],
    bins=[0, 1, 60, 360, 1440, float("inf")],
    labels=[
        "< 1 min",
        "1 a 60 min",
        "1 a 6 h",
        "6 a 24 h",
        "> 24 h"
    ],
    right=False
)

pd.crosstab(
    faixas,
    df["member_casual"],
    margins=True
)
```

```python
pd.crosstab(
    faixas,
    df["rideable_type"],
    margins=True
)
```

```python
df_comparacao = df[
    (df["ride_length_min"] >= 1) &
    (df["ride_length_min"] <= 1440)
]

print("BASE ATUAL")
display(
    df.groupby("member_casual")["ride_length_min"]
      .agg(["count", "mean", "median"])
)

print("\nBASE ENTRE 1 MINUTO E 24 HORAS")
display(
    df_comparacao.groupby("member_casual")["ride_length_min"]
                 .agg(["count", "mean", "median"])
)
```

8.Criando variaveis derivadas

```python
df["day_of_week"] = df["started_at"].dt.day_name()

df["month"] = df["started_at"].dt.month_name()

df["hour"] = df["started_at"].dt.hour
```

```python
df["day_of_week_num"] = df["started_at"].dt.dayofweek

df["year_month"] = df["started_at"].dt.to_period("M").astype(str)
```
